```python
##reading data from keyboard
#input() #by default it takes as str type
a=input("Enter some value for a:")
print(a,type(a))
print("a={} and its type= {}".format(a,type(a)))
print("a=%s and its type= %s" %(a,type(a)))
print("a=%d and its type= %s" %(int(a),type(int(a))))
print(f"value of a= {a} and its type= {type(a)}")
```

```python
#accepting of two numbers and find their product
a=input("Enter value of a:")
b=input("Enter value fo b:")

#converting into our interested format
a=int(a)
b=float(b)
#finding product
c=a*b
#displaying result
print("-"*50)
print("\tvalue of a={} and value of b={}".format(a,b))
print("\n\tProduct of {}*{}={}".format(a,b,c))
print("\n\tProduct of %d * %0.3f = %d " %(a,b,c)) #even we get product as float we get as int
print(f"product of {a} and {b:.6f} is {c:.6f}")
print('-'*50)
```
