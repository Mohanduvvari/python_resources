```python
#Working with anonymous functions or lambda functions
 
# - The anonymous functions are those which doesn't contain any name explicitly 
# - The purpose of anonymous functions is that to perform instant operations
# - The instant operations are those which are used at that point of time only and no-longer interested to use in other parts of the application/project
# - To develop anonymous function we use keyword called 'lambda' and hence anonymous functions are called lambda functions
# - The anonymous function contains single executable statement only but not multiple executable statements
# - The anonymous functions automatically or implicitily return to value no need to use return statement

# syntax> varname=lambda parameterlist:expression

# - Here varname is represents a object of <class function> and hence it is used as anonymous function call
# - Here lambda is keyword used for defining anonymous functions
# - Here paramslist represent list of formal params to store the input comming
# - Here expression represents single executable statement and whose result returns automatically

# example: 
# 1:noraml function
# def sumop(a,b):
#     return a+b

# #mainprog
# res=sumop(10,20)
# print("sum=",res)



# example2:
# 2.lambda function
# addop=lambda a,b:a+b
# #mainprog
# res=addop(10,20)
# print("sum=",res)
```

```python
#example 1: Anonymous functions for cal diff arithmetic operations
import sys 
sumop=lambda a,b:a+b
subop=lambda a,b:a-b
mulop=lambda a,b:a*b
divop=lambda a,b:a/b
floordivop=lambda a,b:a//b
modop=lambda a,b:a%b
expop=lambda a,b:a**b
#main prog
while True:
    print("1.Add\n2.Minus\n3.Product\n4.Division\n5.Floordiv\n6.Modululodiv\n7.Exponential\n8.Exit")
    print('-'*50)
    print("Choose a operation")
    op=int(input("choose an operation:"))
    match(op):
        case 1:
            print("Enter two values for their Addition")
            a,b=float(input("A:")),float(input("B:"))
            res=sumop(a,b) #anonymous function call
            print(f" sum({a},{b})={res}")
        case 2:
            print("Enter two values for their Difference")
            a,b=float(input("A:")),float(input("B:"))
            res=subop(a,b) #anonymous function call
            print(f" Difference({a},{b})={res}")
        case 3:
            print("Enter two values for their Product")
            a,b=float(input("A:")),float(input("B:"))
            res=mulop(a,b) #anonymous function call
            print(f" product({a},{b})={res}")
        case 4:
            print("Enter two values for their Division")
            a,b=float(input("A:")),float(input("B:"))
            res=divop(a,b) #anonymous function call
            print(f" Division({a},{b})={res}")
        case 5:
            print("Enter two values for their floorDivision")
            a,b=float(input("A:")),float(input("B:"))
            res=floordivop(a,b) #anonymous function call
            print(f" FloorDivision({a},{b})={res}")
        case 6:
            print("Enter two values for their remainder/modulo")
            a,b=float(input("A:")),float(input("B:"))
            res=modop(a,b) #anonymous function call
            print(f" remainder({a},{b})={res}")
        case 7:
            print("Enter two values for their ExponentialProduct")
            a,b=float(input("A:")),float(input("B:"))
            res=expop(a,b) #anonymous function call
            print(f" exponential({a},{b})={res}")
        case 8:
            print("Thankyou for your service")
            sys.exit()
        case _:
            print("Your selectin of operation is invalid-TryAgain")
    print('-'*50)
            
```

```python
#using max()
findmax=lambda k,v:max(k,v)
#main program
print("Enter two numerical values:")
a,b=int(input("A:")),int(input("B:"))
maxval=findmax(a,b)
print(f"max({a},{b})={maxval}")
```

```python
#normal function
def findmax(lst):
    maxv=lst[0]
    for val in lst[1:]:
        if val>maxv:
            maxv=val
        else:
            pass
    return maxv

#main prog
#generating list of values:
lst=[]
n=int(input("Enter how many list of items u have:"))
if n<=0:
    print("invalid range!")
else:
    for i in range(n):
        val=int(input("Enter{} value:".format(i)))
        lst.append(val)
res=findmax(lst)
print(f"maxvalue in list {lst} is = {res}")
```
