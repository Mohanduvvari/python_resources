```python
#Functionality of inner loop
for i in range(1,6):
    print('*'*50)
    print(f'outer loop value of i={i}')
    print('*'*50)
    for j in range(1,4):
        print(f'inner loop value of j={j}')
    else:
        print("iam out of inner loop--else part")
        print('_'*50)
else:
    print('iam out of outer loop -else part')
```

```python
#functionality of inner loop using while--while
i=1
while i<=5:
    print('*'*50)
    print(f'outer loop value of i={i}')
    print('*'*50)
    j=1
    while j<=3:
          print(f'inner loop value of j={j}')
          j+=1
    else:
        print("iam out of inner loop--else part")
        print('_'*50)
        i+=1
else:
     print('iam out of outer loop -else part')
```

```python
#generating 1-n multiplication tables 
n=int(input("how many tables you want to create?"))
if n<=0:
    print(f'{n} tables are not possible')
else:
    for i in range(1,n+1):
        for j in range(1,11):
            print(f'{i}*{j}={i*j}')
        print('*'*20)
        
```

```python
#generating multiplication tables from one number to another number
b=int(input("Enter starting number:"))
e=int(input("Enter ending of number:"))
if b<=0 or e<=0:
    print(f'not possible:')
else:
    if b>e:
        b,e=e,b
    for i in range(b,e+1):
        for j in range(1,11):
            print(f'{i}*{j}={i*j}')
        print('*'*20)
```

```python
#validation of student details
while True:
    sno=int(input('Enter sl.no between 100 - 200'))
    if sno>=100 and sno<=200:
        break
    print(f"{sno} is invalid student number")
#validation of name
while True:
    res=True
    sname=input("Enter student name:")
    if sname.isspace():
        print("dont give space try-again")
    elif len(sname)==0:
        print('you must enter student name Try-again')
    else:
        words=sname.split()
        for word in words:
            if not word.isalpha():
                res=False
                break
        if res:
            sname=" ".join(words)
            break
        print(f'{sname},is invalid name try again')
#validation of marks c-100
while True:
    cmarks=input('Enter your C language marks:')
    if not cmarks.isdigit():
        print(f"{cmarks} is not valid")
    else:
        if int(cmarks) in range(0,101):
            break
        print(f"{cmarks} are out of range")
#validation of marks cpp-100
while True:
    cppmarks=input('Enter your CPP language marks:')
    if not cppmarks.isdigit():
        print(f"{cppmarks} is not valid")
    else:
        if int(cppmarks) in range(0,101):
            break
        print(f"{cppmarks} are out of range")
#validation of marks Python-100
while True:
    pymarks=input('Enter your Python language marks:')
    if not pymarks.isdigit():
        print(f"{pymarks} is not valid")
    else:
        if int(pymarks) in range(0,101):
            break
        print(f"{pymarks} are out of range")
```
