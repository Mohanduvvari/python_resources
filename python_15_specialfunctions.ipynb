```python

# #special functions in python

# In python we have 3 special functions they are
# 1>filter()
# 2>map()
# 3>reduce()
# ----------------------------------------------------------------
# 1.filter()
# - This is used for "filtering out some elements form list of elements by applying to function
# - syntax: varname=filter(functionname,iterobj)

# - Here varname is object of type <class 'filter'> and we can convert into any iterobj by using type casting function.
# - functionname represents either normal function or anonymous function
# - The execution process of filter() is that "each value of iterobj will be filtered"
# if the function returns false then that element will be neglected/not filtered this process will be continued until all elements of iterobj completed. 
```

```python
#ex-1> +ve values
def pos(value):
    if value>0:
        return True
    else:
        return False

#main program
print("Enter list of values separated with (,)")
lst=[float(val) for val in input().split(',')]
print(lst)
posval=filter(pos,lst) #here posval is an obj of filterclass
print(posval,type(posval))
poslst=list(posval)
print("list of +ve elements=",poslst)
```

```python
#using lambda
pos=lambda val:val>0
#main program
print("Enter list of values separated by (,)")
lst=[int(val) for val in input().split(',')]
print(lst)
pslist=list(filter(pos,lst))
print("list of +ve elements=",pslist)
```

```python
print("Enter list of values separated with(',)")
lst=[float(val) for val in input().split(',')]
print(lst)
pslst=list(filter(lambda val:val>0,lst))
print(pslst)
```

```python
#obtain list of words whose length ranges between 3-4
print("Enter list of words separated by(',)")
words=[word for word in input().split(',')]
print(words)
word3_4=list(filter(lambda word:len(word)>2 and len(word)<=4,words))
print(f"3-4 length range words form {words} are {word3_4}")
```

```python
#2.map() 
# - Used for obtain new iterobj form existing iterobj by applying old iterobj elements to the function
# - in otherwords map() is used for obtaining new list of elements from existing list of elements by applying old list elements to the function
# - syntax: varname=map(functionname,iterobj)

# - Here varname is obj of map class, and we can convert into any iterobj by using type casting
# - functionname is either normal or anonymous function
# - iterobj is sequence of list,set,dict type 
# - the execution process of map() is that map() sends every element of iterobj to the specified function, this process the element and returns the modified value as result and new list of elements will be obtained this process will be continued untill all elements of iterobj completed
```

```python
#example1> read list of old salaries and give 50% hike to every employee
def hike(salary):
    return salary+(50/100)*salary
print("Enter list of salaries separated by ',' ")
oldsal=[float(sal) for sal in input().split(',')]

salmap=map(hike,oldsal) #here salmap is obj of mapclass
newsal=tuple(salmap)
print("oldsal\t\tnewsal")
for os,ns in zip(oldsal,newsal):
    print(f"{os}\t\t{ns}")
```

```python
#reduce()
 
# - it is used to obtaining a single element or result from given iter-obj by applying to a function
# syntax: varname=reduce(unctionname,iteroobj)
# - here varname is obj of int,float,bool,complex, or str only
# - the reduce() belongs to a pre-defined module called functools

# -Internal flow of reduce()
# -step1> initially reduce() selects first two values of iterobj and place them in first var and second var
# -step2> the functionname lambda,normal utilizes the values of first var and second var and applied to the special logic and obtaines the result
# -step3> reduce() places the result of function in first var and reduce() selects the succeeding element of iterobj and place in second var
# -step4> Repeat step-2,step-3 until all elements completed in iterobj and returns the result of first variable
```

```python
#ex1: 
import functools
def sumop(k,v):
    return (k+v)
lst=[10,20.50,30,40,50]
res=functools.reduce(sumop,lst)
print("sum=",res)
print(type(res))
```

```python
#method 2
import functools
lst=[10,20,30,40,50]
res=functools.reduce(lambda x,y:x+y,lst
                     )
print("sum=",res)
print(type(res))
```

```python
#ex3> max elements from list of elements
import itertools
def mkmax(a,b):
    return a if a>b else b
def min(a,b):
    return a if a<b else b
#main prog
print("Enter alist of values separated with , ")
lst=[float(val) for val in input().split(',')]
print("given list",lst)
maxval=functools.reduce(mkmax,lst)
minval=functools.reduce(min,lst)
print("maxval=",maxval)
print("min value=",minval)
```
