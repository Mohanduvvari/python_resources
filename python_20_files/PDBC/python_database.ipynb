```python
                #types of Queries in database software
# - In RDBMS softwares, SQL queries are classified into 3 types they are
# 1.DDL (Data definition language) create,alter, drop
# 2.DML (Data manipulation language) insert, update, delete
# 3.DQL (Data query language) select

# 1.DDl: The purpose of DDL is that to deal with physical level tables such as table creation with columns, droping tables, re-structure columns of tables. eg: create,alter,drop
#     1.create:
#         this is used for creating tables in database software.
#         eg:create table tablename(col1 datatype, col2 datatype,....)
#         ex: create table emp(eno number(2) primary key, ename varchar2(10) not null, marks number(5,2) not null);
#     2.alter: add, modify 
#         this query is used for altering structure.
#         in other words alter is used for modifying the columnsize(modify) and add new column name (add)
#         eg: alter table tablename modify(existing colname datatype,...)
#             alter table tablename add(newcolname datatype,...)
#     3.drop:
#         this query is used for removing or droping the table from database software
#         eg: drop table tablename
        
```
