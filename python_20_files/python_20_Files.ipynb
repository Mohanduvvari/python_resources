```python
#           Files
#Types of Applications in files
#   
# - The purpose of files in any programming language is that "to maintain data persistency"
# - the purpose of storing the data permanently is called data persistency
# - in this context we can develop two types of applications they are:
#         1. non-persistant applications
#         2.persistant applications
# non-persistant applications development:
# ---------------------------------------
#   - we read the data from keyboard, stored in  main memory (RAM) in the form of     objects processed and whose results displayed on monitor.
#     ex: the programs we did till now comes under non persitant applications
#   - we know that data stored in main memory is temporary.
# persistant applicaations development:
# -------------------------------------
#     - we read the data from keyboard and stored in main memory in the form of object processed and whose results stored permanently. 
#     - in industy wehave two ways to store the data permanently they are:
#         1.by using files
#         2.by using RDBMS database s/w's (oracle,mysql,mongodb,db2,postgreSQL,sqlserver,sqlite3....)
    
```

```python
#Data persisstency by using files of python
#  
# Defination:
# -----------
#         - A file is a collection of records(flat files)
#         -files resides in secondary memory (Hard disks)
#         - Technically A file name is considered as a named location in secondary memory.
#         - The purpose of files is that to get data persisency.
# - All the objects data of main memory becomes records in file of secondary memory and records of file of secondary memory becomes the objects in main memory

# Definition of Stream:
# ---------------------
#     The flow of data between object(s) of main memory and records files of secondary memory is called stream

# ------------------------------------------------------------------------------------
#             Types of files in python
# -------------------------------------------------------------------------------
#  there are two types of files they are:
#             1.text files
#             2.binary files
# 1.text files:
# ------------
#         - A text file is one which always contains alphabets,digits, and special symbols.
#         - Text files are denoted with a letter 'l' (by default)
#         - by default python programming language treats every file as text file 
#         ex: .py, .jav, .cpp, .c, .txt, .xlsx, .doc .... etc

# 2.binary files:
# ---------------
#         - A binary file is one which contains the dat in the form of binary format (pixels)
#         - Binary files denoted by a letter "b"
#         ex: images (.png, .jpeg, .jpg, .gif .....)
#             audio and video (.mvi, .avi, .mp3 ....)
#             pdf file with images
```

```python
#operations on files
 
# on the files we can perfom two types of operations  they are 
#     1.write operations
#     2.read operations

# 1.write operations:
# -------------------:  - the purpose of write operation is that totransfer or save the oject data of main memory as record in the file of secondary memory

# steps: 
#     1.choose the file name
#     2.open the file name in write mode
#     3.perform cycle fo write operations
# - while  we are perfoming write operations we get the following exceptions 
#     a.IOError
#     b.OSError 
#     c.FileExistError

# 2.Read Operations:
# ------------------
#              - The purpose of  read operation is that "to transfer of read the record from file of secondary memory into the object of main memory"
#         steps:
#                 1.choose the file name
#                 2.open the file name in read mode
#                 3.perform cycle of read operations
            
#             - while we are performing read operations we get the following exceptions.
#                 1.FileNotFoundError
#                 2.EOFError  (end of file)
```

```python
#         File opening modes

# - The purpose of file opeining modes is that to open a particular file with the specified mode.
# - in otherwords file opening modes makes us to understand in which mode we are opening the file.
# - in python programming we have 8 file opening modes they are
#         1.r
#         2.w
#         3.a
#         4.r+
#         5.w+
#         6.a+
#         7.x
#         8.x+
# ---------------------------------------------------------------------
# 1. r (read)
# ---------------------------------------------------------------------
# - this mode is used for opening the specified file name in read mode and we perform read operations only 
# - if we dont specify any file operation mode then by default pvm treated 'r' mode as default file mode
# -if we open any file name in 'r' mode and if that file name does not exist in secondary memory then we get FileNotFoundError.

# ---------------------------------------------------------------------
# 2. w (write)
# ---------------------------------------------------------------------
# - this mode is used for creating new file and opens in write mode
# - if we choose new file and the mode is 'w' then pvm creates new file and opened in write mode and we can perform write operations only
# - if we choose existing file and the mode is 'w'  then pvm opens existing file in write mode and existing data of existing file overlapped with new data (old data will gone)

# ---------------------------------------------------------------------
# 3. a (append)
# ---------------------------------------------------------------------
# - This mode is used for creating new file and opens in write mode
# - if we choose new file and the mode is 'a' then pvm creates new file and opened in write mode and can perform write operations only
# - if we choose existing file and the mode is 'a' then pvm opens existing file in write mode and existing data append with new data 

# ---------------------------------------------------------------------
# 4. r+ (read and write)
# ---------------------------------------------------------------------
# - this mode is used for opening the specified file name in read mode and first we perform read operations and later we can perform write operation.
# - if we open any file name in "r+" mode and if that file name does not exist in secondary memory then we get FileNotFoudError.

# ---------------------------------------------------------------------
# 5. w+ (write and read)
# ---------------------------------------------------------------------
# - This mode is used for creating new file and opens in write mode.
# - if we choose new file and the mode is "w+" then pvm creates new file and opened in write mode, and first we can perform write operations and later we perform read operations also.
# - if we choose existing file and the mode is "w+" then pvm opens existing file in write mode and existing data of existing file overlapped with new data

# ---------------------------------------------------------------------
# 1. a+ (write and read)
# ---------------------------------------------------------------------
# - This mode is used for creating new file and opens in write mode
# - if we choose new file and the mode is "a+" then pvm creates new file and opened in write mode, and first we can perform write operations and later we can perform read operations also.
# - if we choose existing file and the mode is "a+" then pvm opens existing file in write mode and existing data of existing file appended with new data

# ---------------------------------------------------------------------
# 1. x (exclusively write)
# ---------------------------------------------------------------------
# - This mode is used for creating new file exclusivly in write mode only once, and we can perform write operations only
# - if we open existing file in "x" mode then we get FileExistError

# ---------------------------------------------------------------------
# 1. x+ (exclusivly write and read)
# ---------------------------------------------------------------------
# - This mode is used for creating new file exclusively in write mode only once and first we can perform write operations and later we can perform read operations also.
# - if we open existing file in "x+" mode then we get FileExistError
```

```python
#               Number of Approaches to open the files
# ----------------------------------------------------------------------------------
#in python programming we have two approaches they are:
#                     1.by using open()
#                     2.by using "with open() as"
# ----------------------------------------------------------------------------------
#1. By using open()
# -----------------------------------------------------------

# syntax: varname=open("filename","file mode")
# - Here varname represents file pointer which will point to the opended file and whose type is  <class , _io.textIOWrapper>
# - open() is used for opening the filein specified file mode
# - filename represents name of the file to be specified by the programmer
# - file mode represents any one of the file opening modes (r,w,a,r+,w+,a+,x,x+)
# - once we open any file name with open() then we  must close the file by using   close() and it is mandatory for the programmer to maintain consistency of data (known as manual closing --> there is no concept of auto closeability in this syntax)

# # ----------------------------------------------------------------------------------
# #1. By using " with open() as"
# # -----------------------------------------------------------
# syntax: with open("filename","filemode") as varname:
#                                 . 
#                                 . 
#                                 block of statements - perform file operatins
#                                 . 
#                 other statements in program

# explanation: 
# - Here "with" and "as" are keywords
# - here varname represents file pointer which will point to the opened file and whose type is  <class, _io.TextIOWrapper>
# - open() is used for openining file in specified file mode
# - file name represents name of the file to be specified by the programmer
# - file mode represents any one of the file opening modes (r,w,a,r+,w+,a+,x,x+)
# - the execution process of "with open() as " is that "as long as pvm represent in side of " with open() as "indentation then file name is activly available and once pvm comes cut off with with open() as indentation then file name is closed automatically and this facility is called auto closeability of file, there is no need to close the file by using close() manually
```

```python
#example - 1. demonistration how to open the file

try:
    fp=open("test1.data")
except FileNotFoundError:
    print("file does not exist")
else:
    print("file opened in read mode and we can perform read operations")
    print("type of fp=",type(fp))
finally:
    print("------------------------")
    print("iam from finally block")
    try:
        print("still, is file closed=",fp.closed) 
        print("----------after closed() -------------")
        fp.close()
        print("still, is file closed=",fp.closed)
    except NameError:
        print("file name itself not opened, there is no question of closing the file")
    finally:
        print("Thx for choosing files")
```

```python
#pickling (object serialization)
 
# - Let us assume there exists an object in main memory which contains multiple values to save or write an object data of main memory into the file of secondary memory by using write() and writeline() they transfer the values in the form of value by value and it is one of the time consuming process (bcoz of multiple write operations)
# - to overcome this time consuming process we must use the concept of pickling.
# - the advantage of pickling concept is that with single write operation, we can save or write entire object data of main memory into the file of secondary memory.

# definition:
# -----------
#             the process of saving or transfering entire object content of main memory into the file of secondary memory by performing single write operation is called pickling 
#             -pickling concept participates in write operations.
# steps for implementing pickling concept:
# ----------------------------------------
#         - import pickle module, here pickle is one of the pre-defined module
#         - choose the filename and open it into write mode
#         - create an object with collection of values (iterobj)
#         - use the dump() of pickle module, dump() save the content of any object into the file with single write operation.
#         - syntax: pickle.dump(iter-obj, filepointer)
#         **note: that pickling concept always takes the file in binary format
```

```python
#un-pickling (De-serialization)
# --------------------------------------- 
# - Let us assume that there exists a record with multiple values in the file of secondary memory. to read or transfer the entire record content from file of secondary memory. if we use read(), readlines() then they read record values in the form of values by values and it is one of the time consuming process (bcoz of multiple read operations.)
# - to overcome this consuming process, we must use the concept of un-pickling. 
# - the advantage of un-pickling is that with single read operation, we can read entire record content from the file of secondary memory into the object of main memory.

# Definition:
# -------------
#     - the process of reading or transfering the entire record content from file of secondary memory into the object of main memory by performing single read operation is called un-pickling.
#     - un-pickling concept participates in read operations.

# steps for implementing un-packing:
# ------------------------------------
# - import pickle module
# - choose the file name and open it into the read mode.
# - use the load() of pickle module. load() is used for transfering of loading othe entire record content from file of secondary memory into object of main memory.
# - syntax: objectname=pickle.load(filepointer)
# **note: un-pickling concpet always takes the file in binary format.
```

```python
#program for copying the content of one file into another file
def filecopy():
    try:
        srcfile=input("Enter source file:")
        with open(srcfile,"r") as sfp:
            srcfiledata=sfp.read() #read data from source file
            dstfile=input("Enter destination file")
            with open(dstfile,"w+") as dfp:
                dfp.write(srcfiledata)
                print("file copied successfully")
    except FileNotFoundError:
        print("Source file not existed")
#main prog
filecopy()
```

```python
#working on CSV files in python
# ---------------------------------------
 
# - CSV stands for comma separated values
# - CSV file is one of the simple format used to store tabular data such as a speadsheet or database
# - A CSV file stores tabular data (numbers and text) in plain text
# - Each line of csv file is a data record. Each record consists of one or more fields spearated by commas
# - python provides an in-built module called csv to work with csv files.
# - there are 2 classes provied by this module for writing class object
#         1.By using Usingcsv.writer class obj
#         2.By using csv.DictWriter class obj
# - these are 2 classes provided by this module for reading the data from csv file they are
#         1.By using Usingcsv.reader class obj
#         2.By using csv.DictReader class obj


# #By usingcsv.writer class obj:
# # --------------------------------
# - The csv.writer class obj is used to insert data to the csv file.
# - to create an obj of csv.writer class obj we use writer() and present in csv module
# - "csv.writer" class obj provides two functions for writing to csv file
# - they are :
#                 1.writerow()
#                 2.writerows()
#         1.writerow(): this method writes a single row at a time.field row can be written using this method.
#                 syntax: csvwriterobj.writerow(fields row / data row)
#         2.writerows(): this method is used to write multiple rows at a time. this can be used to write rows list.
#                 syntax: writing csv file in python
#                         csviterobj.writerows(data rows)
#                 here data rows can be list tuple,set,frozenset only

# #2.By using csv.DictWriter class obj:
# # -------------------------------------
# - the csv.Dictwriter class obj is used to insert dict data to the csv file.
# - To create an object of csv.Dictwriter class obj we use Dictwriter() and present in csv module.
# - csv.Dictwriter class object provides two functions for writing to csv
#                 1.writeheader()
#                 2.writerows()
# 1.writeheader():
# -----------------
# - writeheader() method simply writes the first row of your csv file using the pre-specified fieldnames.
# - syntax:Dictwriterobj.writeheader()
# ----------------
# 2.writerows():
# -----------------
# - writerows() method simply writes all the values of (key,value) from dict object in the form of separate rows [note: it writes only the values (not keys)]
# - syntax: Dictwriteobj.writerows(dictobj)

# -----------------------------------------------------
#                 Reading the Data from csv file
#                 -------------------------------
# - there are various ways to read a csv file that uses either the csv module or the pandas library.
# - the csv module provides classes for reading information from csv file
#                 1.csv.reader
#                 2.csv.DictReader
# ----------
# 1.csv.reader()
# --------------
# - this function is used for creating an object of csv.reader class and it helps us to read the data records from csv file
# - syntax: csvreaderobj=csv.reader(filepointer)

# -------------------------
# 2.csv.DictReader():
# ------------------------
# - this functin is used for creating an obj of csv.DictReader class and it helps us to read the data from csv file where it contains dict data (key,value)
# - syntax: csvdictreaderobj=csv.DictReader(filepointer)  
```

```python
#program for creating the csv by using writer() of csv module
import csv #step-1
cols=['sno','name','marks']#step-2
records=[[100,'aniket',66.66],
         [101,'Rajesh',65.66],
         [102,'Rudra',77.77],
         [103,'kvr',11.11],
         [104,'rishi',55.55] 
         ] #step-3
with open("student.csv",'w') as fp: #step-4
    csvwr=csv.writer(fp) #here csvwr is an object of <class, csv.writer> #step-5
    #here csvwr contains two functions 1.writerow() 2.writerows()
    csvwr.writerow(cols) #step-6
    csvwr.writerows(records) #step-7
    print("csv file created successfully -- verify")
 
```

```python
#program for creating the csv file by using Dictwriter of csv module
import csv #step-1
cols=['pno','pname','price'] #step-2
records=[{"pno":111,'pname':'kitkat','price':20},
         {"pno":222,'pname':'cadburry','price':30},
         {'pno':333,'pname':'chacopie','price':30},
         {'pno':444,'pname':'bingo','price':30},
         {'pno':555,'pname':"sting",'price':40}
         ]
with open("product.csv","w") as fp:
    csvwr=csv.DictWriter(fp,fieldnames=cols) 
    #here csvwr contains two functions 1.writeheader() 2.writerows()
    csvwr.writeheader()
    csvwr.writerows(records)
    print("CSV file created successfully -- verify")
```

```python
#program for reading data from csv file by using _io.textiowrapper
with open("product.csv",'r') as fp:
    records=fp.read()
    print('-'*50)
    print(records)
    print('-'*50)
```

```python
#program for reading the data from csv file by reader() in the form list
import csv
with open("product.csv",'r') as fp:
    csvr=csv.reader(fp) #here csvr is an obj of _csv.reader
    for records in csvr:  #here records is list type
        for val in records:
            print(val,end="\t")
        print()
    
```

```python
#program for reading the data from csv file by dirctreader()  in the form of dict
import csv
with open("product.csv",'r') as fp:
    print("--------------------------")
    csvr=csv.DictReader(fp) #here csvr is an object of csv.DictReader
    for record in csvr: #here record is dict type
        for k,v in record.items():
            print("\t{}-->{}".format(k,v))
        print("---------------")
    print("------------------------")

```

```python
#working with json files
# ----------------------------
  
# - json stands for javascript object notation
# - json file always contains an extension.json(filename.json)
# - json file format is one of the light weight file format which is used in web application development for exchange the data between client (browser) to server side application.
# - file format contains the data in the form of (key,value) and in python it is related to dict obj
# - to work with json operations we use a pre-defined called 'json'
# - when we work on json concept we have two types of json data they are:
#         1.json data
#         2.json file dat


# 1.json data:
# --------------
# - syntax:jsondata='{"key1":'val1','key2':'val2',......,'key-n':'val-n'}'
# - to conver json data into dict data we use loads() which is present in pre-defined module called json.

# ---------------
# 2.Dict data to json data
# -------------------------
# - syntax: Dictobj={key1:val1,key2:val2,....,key-n:val-n}
# - to convert dict data into json data we use str()

# ---------------
# 3.Dict data into json file data
# ---------------------------------
# - syntax: Dictobj={key1:val1,key2:val2,key3:val3,.....,key-n:val-n}
# - to save the dict obj data into the json file, we use dump() of json module

# -----------------------
# 4.json file data into dict data
# -----------------------
# - to read/transfer json file data into dict data we use load() of json module
# - syntax: Dictobjdata=json.load(file pointer)
```

```python
# program for converting dict obj data into json string data by using str() 
dictobj={'sno':123,'name':'Mohan','marks':89.50}
print("-"*50)
print("content of dictobj=",dictobj)
print("type of dictobj=",type(dictobj))
print("-"*50)
#convert dict obj data into json string data
jsonstrfrmt=str(dictobj)
print("content of json string format=",jsonstrfrmt)
print("type of json string format=",type(jsonstrfrmt))
print("-"*50)
```

```python
#program for saving the dict obj into the json file by using dump() of json module
import json
dictobj={'sno': 123, 'name': 'Mohan', 'marks': 89.5}
print("-"*50)
print("content of dictobj=",dictobj)
print("type of dictobj=",type(dictobj))
print("-"*50)
#saving the dict data into the json file
with open("student.json",'w') as fp:
    json.dump(dictobj,fp)
    print("Dict data saved in json file as dict format -- verify")
```

```python
#program for   converting json file data into dict obj data by using load() fo json module
import json
try:
    with open("student.json",'r') as fp:
        dictobj=json.load(fp)
        print("Content of dictobj=",dictobj)
        print("Type of dictobj=",type(dictobj))
        print("-"*50)
        for k,v in dictobj.items():
            print("\t{}--{}".format(k,v))
        print("-"*50)
except FileNotFoundError:
    print("Json file data does not exits")
```

```python
#program for converting json string data into dict obj by using json.loads()
import json
dic={'sno': 123, 'name': 'Mohan', 'marks': 89.5}
jsonfrmt=json.dumps(dic) #dumps converts python obj into json string

# json.dumps() and json.dump() both convert python objects into json format but the difference is where the json data goes
# json.dump() used to write json data directly into file
#json.dumps() used to convert python obj into json string

#similary loads() read json form file and load() convert json str into python obj


# jsonfrmt='{"sno":"122","name":"Mohan","marks":"89.90"}'
#converting jsonstrformat into dictionary
print("-"*50)
print("Content of json string format=",jsonfrmt)
print("type of json string format=",type(jsonfrmt))
print("_"*50)
#convert json str data into dict
dictobj=json.loads(jsonfrmt)
print("Content of dictobj=",dictobj)
print("type of dict obj=",type(dictobj))
print("_"*50)
for k,v in dictobj.items():
    print(f"{k}-->{v}")
print("_"*50)
```

```python
#working os based operations with files and folders
#  
# - we know that os is one of the resource (main memory, secondary memory, processor, input output devices).


# - Allocation and De-allocation manager during program execution
# - with os we can perform the followin operations
# - create a folder --- mkdir(filename)
# - delete a folder --- rmdir(filename)
# - remove a folder/file -- remove(file name)
# - rename a folder/ file name -- rename(file name/ folder name)
# - listing the files in folder,--- listdir()


# - to work with above operations we use pre defined module called 'os'

# - folder is one of the file name and it does not take memory space in secondary memory (HDD)
# -  file takes memory space in secondary memory (HDD)

# 1>creating folder/directory:
# -----------------------------
#     - for creating a folder we use mkdir()
#     - syntax: os.mkdir("Foldename")
    
#     - if the folder name already exist then we get FileExistError
#     - mkdir() can create only one folder at a time and if we try to create folders hierarchy then we FileNotFoundError

# 2>delete floder:
# -----------------
#     - to delete folder we use rmdir()
#     - syntax: os.rmdir("folder name")
#     - rmdir() can remove folder names provided folder name is empty
#     - if we specify any folder name with escape sequence(\n \u \ digits \ t etc) then we get osError
    
# 3>remove file:
# -------------------
#     - to remove file we use remove(filename)
#     - syntax: os.remove("absolute path of file name )
#     - if the file name does not exist then we get FileNotFoundError

# 4>rename folder/file :
# ---------------------
#         - to rename a folder, we use rename()
#         - syntax: os.rename("oldfilename","newfilename")
#         - syntax: os.rename("oldfoldername","newfoldername")
#         - if the old foldr name does not exist gives FileNotFoundError
        
# 5>list the file names in folder:
# ------------------------------------
#         - to list the file names in folder we use listdir()
#         - syntax: os.listdir("absolute path of folder name")
#         - if the folder not exist gives FileNotFoundError
```

```python
#Program for creating folder/dirctory
#mkdir()
import os
try:
    os.mkdir("test1")
    print("\tfolder created successfully -- verify")
    os.mkdir("test2")
    print("\tfolder created successfully -- verify")
except FileNotFoundError:
    print("mkdir() can create only one folder at a time!")
except FileExistsError:
    print("\tthe specific folder already exist")
except OSError:
    print("\tcheck ur path of folder names")
```

```python
#program for deleting folder
import os 
try:
    os.rmdir("test1")
    print("\tfolder deleted successfully")
except FileNotFoundError:
    print("\t given folder not existed")
# except OSError :
#     print("\trmdir() can remove those folder which are empty -- check patha")
except Exception as e:
    print(e)
```
