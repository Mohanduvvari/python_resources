```python
#AND/and
#exp1 and exp2
# t t t
# t f f
# f t f
# f f f
```

```python
#examples #basically result is true if both exps are true, if any one if false  result is false/ if the first exp will be false then pvm will not check the second exp coz it is waste of time, if we have 3 expressions 1st is true and second is false pvm decide result is false (This checking method is called short circuit evalution)
print(10>3 and 40>25) # t  t ->  t
print(10>3 and 40<25) # t  f ->  f 
print(10<2 and 10>3)  # f  t ->  f
print(10<3 and 10<7)  # f  f ->  f
```

```python
# to check how the short circuit evalution was working
print(100 and 200) #we give T and T so pvm check first is true or not it is true so pvm will check for second value it is also true so result is decided by second true i.e 200 (pvm checking till last exp is called full circuit evalution)
print(100 and 0 and 10000) #here even the 1st one is true the second is false so result is decided by 0 i.e false (pvm decides its result without go to last exp then this is called short circuit evalution)
print(None and 1 and 2)  #here the result is decided by 1st item F T T
print(0 and 1 and 2) # F F F
```

```python
# 'OR' operator (any one of expression is true then entire expression or compound expression is True)
# exp1 OR exp2
# T T T
# T F T
# F T T
# F F F
```

```python
print(10>3 or 40>25) # t  t ->  t
print(10>3 or 40<25) # t  f ->  t 
print(10<2 or 10>3)  # f  t ->  t
print(10<3 or 10<7)  # f  f ->  f
```

```python
#short circuit evalution
#in or operator if first exp is true then entire exp is true
#if first exp is false then pvm will check for second exp if it is also false then check for next exp if the next is True then the entire compound exp is becomes true
print(10 or 0) #10 decides the result 
print("" or None or 0 or 0b000 or 1 or False) #here result is decided by 1 which is at 5th place
```

```python
#bitwise operators
#purpose is to perform bit by bit operations 
#the bitwise operators are applicable on only integer data but not on floating point coz floating values does not have certainity like int data 

#Execution process
#1. bitwise operatior convert given int to binary value
#2. depends on type of bitwise operator, pvm convert binary to decimal automatically as python is highlevel programming lang
#in programming we have 6 types
# 1. << bitwise leftshift
# 2. >> bitwise rightshift
# 3. & bitwise AND 
# 4. | bitwise OR operator
# 5. ~ bitwise compliment
# 6. ^ bitwise XOR
```

```python
#1. bitwise leftshift <<
#the execution process of bitwise leftshift is , that moves no.of bits to left side by adding no.of zeros at right side 
a=10
b=a<<3 #for our understandig we have formula as value*2^no.of bits
print(f"for the bitwise left shift of {10} by {3} bits we get {b}")
print('-'*50)
print(a*(2*2*2)) 
print(10*pow(2,3))
```
