```python
#Parameters and Arguments


#parameters 

#parameters are basically two types:
#  1.formal parameters/variable
#     ->used in function heading
#     ->the use of formal parameters is that to store the input comming from funcall
#  2.local parameters/variables
#     ->used in function body
#     ->used for local parameters/variables is that to store the temparely        #         results of functin processing result
#the values of formal parameters and local parameters can accessed within function definition only but not possible to access in other parts of the program

#example: def sumop(a,b,c): #a,b,c are called formal params
#                   d=a+b+c
#                   e=a*b*c  #here d,e are called local variables


#arguments (arguments or parameters)
#arguments are nothing but variables or values used in function call

#example:
#a=10
#b=5
#c=7
#sumop(a,b,c) #function call  (a,b,c are called arguments)
# or
#sumop(10,5,7)  10,5,7 are called argument values
# 
#note: the relationsip between arguments and parameters is that 'every value of argument(s) from function call passing to parameter(s) in function definition this process is called argument passing mechanism
# 
# Types of arguments or parameters in python
# ->based on passing the arguments values from function call to formal parameters of function definition the arguments are classified into 5 types
# 
# 1.possitional arguments
# 2.Default arguments
# 3.Keyword arguments  
# 4.Variable length arguments
# 5.Keyword Variable Length arguments
```

```python
#positional arguments:

# -->positional arguments mechanism is that default pasing arguments mechanism used     by pvm for passing the values of arguments of functions call, to formal parameters of functin definition

# -->postitional arguments concept says that every arguments values passing to every formal parameter based on their position by maintaining order and meaning for heigh accuracy in oterwords the number of arguments must be equal to number of formal parameters

# -->PVM gives high priority for positional arguments

#    eg: def funname(param1,param2,param3.... ,param-n):
#                 block of statements perform operations
       
#        funname(arg1,arg2,arg3,......arg-n)
```

```python
#Default parameter or arguments:


# when there is common values for familty of similar function call then such type of common values must be taken as default parameters with common values (but not recommended to pass by using positional argumant or parameteres)

# syntax:
#            def funname(param1,param2,.....param n-1=val1,param n=val2):
#                         block of statements
            
#             funname(arg1,arg2,arg3......arg n-1,argn)

#     #here param1,param2,param3... called formal params, and param n-1, paramn are called default params
#     # when we use default parameters in the fuction definition they must be used at last of parameters otherwise gives Syntax error
#     #use default params when we dont know the exact argument, so inseted of using exeact value we can use default value 
```

```python
#keyword parameters or arguments


# in some of circumstances we know the function name and formal param names and we dont know the order of formal params and to pass the data/values accurately we must use the concept called keyword parameters or argumens
# -->the implementation of keyword parameters or arguments says that all that the formal paramets names used as arguments in function calls[s] as keys

# syntax: def functionname(param1,param2,....param-n):
#              block of statements

#         functionname(param-n=val-n,param1=val1,param n-1=val n-1......)


# -->param n =val n, and param1=val1 are called keyword arguments
# -->when we specify keyword arguments before possitional arguments in function call then we get syntax error , because pvm gives 1st preference ot positional argument, acutally keyword arguments should be given after positional arguments.
```

```python
#variable length parameters or arguments

# when we have family of multiple similar function calls with variable number of values/arguments then with normal python programmig we must define multiple function definitions this process leads to more development time

# --> to avoid this  process we must use the concept called variable lenth parameters.

# --> to implement variable length parameters concept we must define single function definition and takes a formal paramets prceded with symbol called  astrisk*  and he formal parameter with astrisk symbol called variable lenght parameters and whose purpose is to hold store any number of values comming from similar functin calls, and whose type is <class'tuple'>

# syntax: def funname(list of formal positional params, *params, param2=value2......):
    
# here *params called variable length parameters and it can hold any number of arguments values/variables number of arguments and *params is tuple type

# #rule1:the *params must always written at last part of function heading (when we have positional and variable lenght params only and it must be only one )

# #rule2:  when we use variable length and default parameters in function definition then we must use default parameters at last and before we use variable leght parameters and in functin calls we should not use default parameter as keyword argument in middle of function calls coz variable number of values are treated as positional arguments values and we must write them in last in function calls
# note: sno,sname,*marks=10,'rs', 45, 67,89,90
# print(sno,sname)  --> 10,'rs'
# print(marksntype(marks)) -->[45,67,89,90], list
```

```python
###                  Local and Global variables
#local variables:
""" 
- local variables are used in function body
- purpose  is that to store temp results after function processing
- local variables can be accessed inside of corresponding function definition only but not possible to access other parts of program (this is called local access)
"""

#Global variables:
""" 
- global variables are those which are used for providing common value for all different function definitions (called global access)
- in memory management point of view global variables values takes less memory space
- global variables to be defined before all the function calls, so that we can access global variables values in all those function definitions


def fun1():

def fun2():
def fun3():
#main program
var1=val1
var2=val2

fun1()
fun2()

here var1,var2.... are called global variables and  defined before function calls, so that we can access those values inside of corresponding function definitions
"""


#global keyword

""" 

- when we want modify the global variables values inside of functin definition then global variables names must be preceeded with 'global' keyword otherwise we get unboundLocalError: cannot access local variables when it is not associated with a value

syntax: var1=val1
        var2=val2
        .
        . 
        varn=valn
        def fun():
        
        global var1,var2,var3......varn #modify variables

#note: to modify global variable value inside of function definition we use global keyword (mandatory to write)
# 
# to access global variable value inside f function definition we dont use global keyword        
"""
```

```python
#           global and local variables and globals()
""" 
- when we come accross same global variable names and local variable names in same function definition, then pvm gives preferene for local variable but not global variables
- in this context, to extract/retrieve the values of global variables names along with local variables we must use globals() and it returns an object of <class 'dict'> and its dict obj stores all global variable name as keys and global variable values as values of values 

syntax: var1=val1
        var2=val2
        . 
        . 
        varn=valn

        def fun1():
            var1=val1
            var2=val2
            . 
            . 
            varn= valn

            dictobj=globals()
            global val1=dictobj['var1'] or dictobj.get('var1) or globals().get('var1')
             
            global val2=dictobj['var2'] or dictobj.get('var2') or globals()['var2'] or globals().get('var2')
"""
```
