```python
#Functions in python
#the purpose of function is that to perform certain operation/task and provide code re-usability
#Function:sub program of main program is called function or part of main program is called function

# parts of Functin: 
# 1.function definition
# 2.function call

# -->Function definition exists only once
# -->we can call Function call any number of times
# -->function definition will execute, only when we call by using function call without calling function pvm will not execute functin definition

# -->phases: 1.every function must take an input
#            2.every function must process the input
#            3.every function must give output or return result

# syntax:
#     def functionname(list of formal parameters if any): - function heading
                                
#                                 from statemt-1-n called function body
#                 statement-1 
#                 statement-2
#                 . 
#                 . 
#                 statement-n 

# 1.def is keyword, used to define programmer defined function
# 2.functioname is represented a valid variable name and treated as function name and every function name is an object of type <class:function>
# 3.list of formal parameters represent list of  variable names used in function heading
# 4.statement in function body are called indetation block of statement, process inputs,logic for problem solving and it is known as business logic or function process logic
# 5.in function body we can use some special variable and they are called local variables and whose prpose is to store the temporary results
# 6.the values of formal parameters and local variables can be accessble obly inside correspondig functin definiton but not possible to access in other parts of the program and in oter definitions (this is called scope of formal params,and local vars)
```

```python
#program for finding occurence(s) of each letter or value in given word
def getline():
    return input("Enter a line of text")
def getoccurance():
    line=getline()
    d={}
    for ch in line:
        if ch not in d:
            d[ch]=1
        else:
            d[ch]=d.get(ch)+1
    else:
        print(d)
        print(f'length={sum(d.values())}')

getoccurance()
        
```

```python
#accept a text and convert into lower and upper
#without using lower(),upper()
#lowerconverter
def getline():
    return input("Enter line of text:")
def lowerc(line):
    lc=''
    for ch in line:
        if ord(ch) in range(65,91):
            lc=lc+chr(ord(ch)+32)
        else:
            lc+=ch
    return lc
def upperc(line):
    uc=""
    for ch in line:
        if ord(ch) in range(97,123):
            uc+=chr(ord(ch)-32)
        else:
            uc+=ch
    return uc
#main function
line=getline()
lower=lowerc(line)
upper=upperc(line)
print(f"given lin of text={line}")
print('-'*40)
print(f"lower ==>{lower}")
print(f'upper==>{upper}')
```

```python
#find maxlength of word from given line of text
def getline():
    return input("Enter a line of text:")
#function for finding max length word
def getwordlen():
    line=getline()
    if len(line)==0:
        return {}
    elif line.isspace():
        return {}
    else:
        words=line.split()
        d={}
        for word in words:
            d[word]=len(word)
        return d
def maxwrd():
    d=getwordlen()
    if len(d)==0:
        print("there is no data present in dict and cant find max length word")
    else:
        lst=[]
        ml=max(d.values())
        for w,l in d.items():
            if l==ml:
             lst.append(w)
        print("max length word(s):",end=" ")
        for val in lst:
            print(val,end=" ")
#main program
maxwrd()

```

```python
#list of values and sort them in ascending and decending order
def readvalues():
    n=int(input("Enter how many numbers u have"))
    if n<=0:
        return []
    else:
        list=[]
        for i in range(1,n+1):
            item=int(input(f"Enter value of {i}"))
            list.append(item)
        return list
def sortitems():
    lst=readvalues()
    if len(lst)==0:
        print("list is empty cant'sort data")
    else:
        print("Given Elements")
        print(lst)
        print('-'*40)
        lst.sort()
        print(f"Ascending order:{lst}")
        lst.reverse() #we can also use lst.sort(reverse=True)
        print(f"Decending order:{lst}")

#main program
sortitems()
```

```python
#accepting list of values and find sum and average
def readlist():
    n=int(input("how many values u have:"))
    if n<=0:
        return []
    else:
        lst=[]
        for i in range(1,n+1):
            val=int(input(f"Enter a value of {i}"))
            lst.append(val)
        return lst
    
def sumavg(lst):
    if len(lst)==0:
        print("we cant find  find sum or avg for a empty list")
    else:
        total=sum(lst) #or we can find totoal using loop, never use keywords as varnames
        average=total/len(lst)
        print(lst)
        print("-"*40)
        print(f"sum of list items={total}")
        print(f'Avg of list items:{average}')

lst=readlist()
sumavg(lst)
```

```python
#capitalize the text 
def getline():
    return input("Enter a line of text")
def capitalize():
    line=getline()
    if len(line)<=0:
        print("we cant do operations on empty text")
    else:
        print(line)
        print('-'*40)
        chs=""
        for ch in line[1:]:
            if ord(ch) in range(65,91):
              chs=chs+chr(ord(ch)+32)
            else:
              chs+=ch
        if ord(line[0]) in range(97,123):
            chs=chr(ord(line[0])-32)+chs
        else:
         chs=line[0]+chs
        return chs
cap=capitalize()
print(cap)

```
