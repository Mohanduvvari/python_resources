```python
#comprehensions concepts
 
# - The purpose of comprehensions concept is that to define or develop precise  or concise code

# syntax1> varname=[exp1 for varname in iterobj if test cond]  #list comprehension
# syntax2> varname=[exp1 for varname in iterobj]

# syntax3> varname={exp1 for varname in iterobj if test cond} #set comprhension
# syntax4> varname={exp1 for varname in iterobj}

# syntax5> varname={exp1:exp2 for varname in iterobj if testcond} #dict comprehension
# syntax6>varname={exp1:exp2 for varname in iterobj}
```

```python
#example1> List comprehension
#getting list of +ve elements
lst=[10,-20,30,-40,50,-60,-15,25]

plist=[]
for val in lst:
    if val>0:
        plist.append(val)
print(lst)
print(plist)
print("-"*50)
pslst=[val for val in lst if val>0]
print(lst)
print(pslst)
```

```python
#read values from kwb using list comprehension
print("enter a list of values separated by space:")
inpu=input()
print(inpu)
inpu=inpu.split(" ")
print(inpu)
print("-"*50)
print("enter a list of values separated by space:")
lst=[float(val) for val in input().split()]
print("lst of values:")
print(lst)
```

```python
#prog for getting +ve  elements
lst=[10,-20,30-40,50,-15,-16,25,0,45]
poset={val for val in lst if val>0}
negset={val for val in lst if val<0}
print(lst)
print(poset)
print(negset)
```

```python
#find length of every word from list of words
words=['python','java','DSA','kotlin','c']
wordln={word:len(word) for word in words}
print(wordln)
print('-'*50)
for k,v in wordln.items():
    print(" {}-->{}".format(k,v))
```

```python
#len of every word which contains purely alphabets from list of words
words=['python','java','DSA','kotlin','c','1233','12abx']
wordsal={word:len(word) for word in words if word.isalpha()}
print(wordsal)
```

```python
#find square of every value of list
vals=[12,4,6,18,3,9.1,'a','Arun']
sqrval={val:val*val for val in vals if str(val).isdecimal()}
for val in sqrval.items():
    print(f"{val[0]} --> {val[1]}")
```

```python
#read values from kwb and find sqr and sqrt by using comprehension concept
print("ente a list of values separated by space")
lst=[float(val) for val in input().split()]
print("given list=",lst)
sqr={val:val**2 for val in lst}
sqrt={val:val**0.5 for val in lst}
print('-'*50)
print("square value:")
for val in sqr.items():
    print(f"{val[0]}-->{val[1]}")
print('-'*50)
print("square roots")
for val in sqrt.items():
    print(f"{val[0]}-->{val[1]}")
print("-"*50)
print(f"given value\t square value\t squareroot")
for k,v in sqr.items():
    print(f"{k}\t\t {v}\t\t {round(sqrt[k],2)}")
```

```python
#increase hike of 50% of every employee
oldemsal={100:1.2,200:2.3,300:4.5,500:6.0}
newempsal={eno:round(oldemsal[eno]+oldemsal.get(eno)*50/100,2) for eno in oldemsal}
print("\tempno\toldsalary\tnewsalary")
for eno in oldemsal:
    print(f"\t{eno}\t{oldemsal.get(eno)}\t\t{newempsal.get(eno)}")
```

```python
#traditional programming
#increasing 50% hike to every employee
oldemsal={100:1.2,200:2.3,300:4.5,500:6.0}
print("\tempno\toldsalary\tnewsalary")
for i,j in oldemsal.items():
    print(f'\t{i}\t{j}\t\t{round(j+((50/100)*j),2)}')
```
