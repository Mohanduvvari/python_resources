```python
#capitalize()
st="python is OOP language"
print(st,id(st))
print('-'*40)
s=st.capitalize() #here id will be changed
print(s,id(s))
```

```python
#title() used to obtain title case of a given sentence or making all words first letter as capital
st="python is OOP language"
print(st,id(st))
print('-'*40)
print(st.title())
```

```python
#index('specific value') obtain index of the specific value
#if the specific value does not exist gives value error
st="python is OOP language"
print(st)
print('-'*40)
print(st.index('i'))
```

```python
st="python is OOP language" #gives index of first occurance of specific value
print(st)
print('-'*40)
print(st.index('a'))
```

```python
#upper()
st="python is OOP language"
print(st)
print('-'*40)
print(st.upper())
```

```python
#lower()
st="Python is OOP language"
print(st)
print('-'*40)
print(st.lower())
```

```python
#islower()
st="Python is OOP language"
s=st.lower()  #python is oop language
print(st)
print('-'*40)
print(st.islower())
print(s.islower())
```

```python
#isupper()
st="Python is OOP language"
s=st.upper()   #PYTHON IS OOP LANGUAGE
print(st)
print('-'*40)
print(st.isupper())
print(s.isupper())
```

```python
#isdecimal()  most strict 
#accepts only decimal digits 0-9 
#no unicodes, no subscripts,no fractions, not roman allowd
print('123'.isdecimal()) 
print('²'.isdecimal())
print('⅖'.isdecimal())
print('Ⅳ'.isdecimal())
```

```python
#isdigit() returns true if the given str is purely digits and some digit like unicodes like subscripts and superscripts etc otherwise returns false
#moderatly strict
print('123'.isdigit()) 
print('²'.isdigit())#subscript is allowd
print('⅖'.isdigit())#fractions is not  allowd
print('Ⅳ'.isdigit()) #roman number not allows
```

```python
#isnumerical() returns true if the given str is purely digits otherwise returns false
#flexible 
print('123'.isnumeric()) 
print('Ⅳ'.isnumeric()) #roman number allowd
print('2²'.isnumeric()) #subscript is allowd
print('⅖'.isnumeric()) #fractions is allowd
```

```python
#isalpha() returns true if the given str is purely alphabets otherwise returns false
st='python' 
ss='python world' #here str has space which is not str so false
s='1234567'
print(st.isalpha())
print(ss.isalpha())
print(s.isalpha())
```

```python
#isalnum() returns true if the given str contains alphabets, numbers or both  otherwise returns false
st='ac12%^'
ss='234acdc'
sd='asdfllkjgh'
s='1234567'
s1='python3.13.5' #it contains '.' which is neither alpha nor number
print(st.isalnum())
print(ss.isalnum())
print(sd.isalnum())
print(s.isalnum())
print(s1.isalnum())
```

```python
#isspace()
#true if word contains purely space else false
print("".isspace())
print(" ".isspace())
print("mohan krishna".isspace())
```

```python
 #split()
#splitting the given str obj data into different words based on specific delimeter
#like -,_,#,%,;,......
#default delimeter is space
#output is list type
s='python is an oop language'
sp=s.split() #here default delimeteris space
print(s,id(s))
print(sp,type(sp),id(sp))
```

```python
dob='1-12-2000'
db=dob.split('-') #delimeter is '-'
print(dob)
print(db)
print(db[0])
print(db[1])
print(db[2])
```

```python
#join() #combining or joining list of values from any itr obj
ls=['hyd','bang','ap','chennai']
s=' ' #we can use any delimeter
print(s.join(ls))
print("*".join(ls))
```

```python
#isascii() return true if str has only ascii otherthen ascii gives false
print("mk".isascii())
print("mk1234".isascii())
print("mk@#$".isascii())
print("@#$.@#".isascii())
print("3.13.5".isascii())
print("😂".isascii())
```

```python
#startswith()
#returns true is strobj starts with specific character other then that gives false
print('Python'.startswith('p'))
print('Python'.startswith('P'))
print('Python'.startswith('s'))
```

```python
#endswith()
#returns true is strobj ends with specific character other then that gives false
print('Python'.endswith('p'))
print('Python'.endswith('n'))
print('Python'.endswith('on'))
```

```python
#swapcase()
#make lower to upper and upper to lower 
print("pYTHON".swapcase())
print('PyToHoN'.swapcase())
```

```python
#casefold() #for looking casefold and lower is similar but casefold is very important functin while comparing str in a casesensitive way especially with international texts
print("Mohan krishna".casefold())
print("MOHAN KRISHNA".lower())
print('-'*50)
print('∑'.lower())
print('∑'.casefold())

```

```python
#count()
print('mississippi'.count('i'))
```

```python
#lstrip()#left space removed
#rstip() #right space removed
#strip()
print("          mohan")
print("          mohan".lstrip)
print("      mohan                   ".rstrip())
print("          mohan              ".strip())
```

```python
#replace()
#replace old val with new val
print("Python".replace('P',"M"))
```

```python
#__contais__()
print("hello world".__contains__('ll'))
print("hello world".__contains__('ok'))
```
