```python
#function
#------------------------
#  
# Function:
# ----------------------
# - A function is one of the subprogram of main prog or a part of main prog is called function
# - A function is a subprogram, which is used to perform certain operation and provides code re-usability
# - A program related code can be accessed within same program but not possible to access accross the program

# Module:
# ----------------
# - A module is a collection of global variable functions, classnames
# - the purpose of module is that to re-use the global vars, functions, classnames of module either the same program or accross the programs provided module and its corresponding provided module and its corresponding main program must prasent in same folder but not possible to access accross the folders or environments or networks

# package:
# -----------------
# - A package is a collection of modules
# - The purpose of packages is that to re-use modules(global vars,functions, classes) either within same folder or different folders or environment or networks
# *note: The concept of functions, modules and packages are called re-usable techniques in functional programming 
```
