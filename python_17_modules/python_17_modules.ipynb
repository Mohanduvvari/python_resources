```python
#python modules:
#------------------------- 
# - We know that functions are used for performing certain operations and provides code-reusablility withing the same program but not able to provide re-usablility accross the programs
# - the purpose of modules concept is that to reuse the functions, global variables and classes from one program to another program provided both the programs present in the same folder

# def of modules:
# -----------------
# a module is a collection of global variables names, functions and class names

# types of modules:
# -----------------
# 1>predefined modules
# 2>programmer or user defined/ custom defined modules

# 1> pre-defined modules:
# -----------------------------------
# -these modules are already defined by python lang developers and available in python s/w and used by all python lang programmers and for deling with universal requirements
# ex: functools,sys,math,cmath,calander,re,pickle,threading,csv... etc.
# out of many pre-defined modules by default one of the pre-defined module called 'builtins' imported to all python programs and it is called default imported python module

# 2> programmer or user or custom defined modules>
# -----------------------------------------------------
# - these modules are developed by python programmer and available in python project and used by other members of same project for dealing with common requirements
# ex: addition,mathsinfo,icici...................


# Development of programmer defined modules
# --------------------------------------------
# to develop programmer defined modules we must use the following steps:
#     1>define global variables
#     2>define functions
#     3>define classes

# - After developing step2,step3 we must save with some file name with extension '.py'
#   (filename.py) and it is treated as modulename-->filename=module name

# - Hence if the python file contains global var name, function names&class names and it is treated as module name

# - when a file name treated as a module name, internally python execution environment creates a folder automatically on the name of __pycache__ and it contains module name on the name of 'filename.cpython-313.pyc'

# ex: __pycache__      -->folder name
#         |______Aop.cpython -313.pyc  ------->module name
#         |______Mathsinfo.cpython-313.pyc ---->module name
#         |______icici.cpython-313.pyc     ---->module name
```
