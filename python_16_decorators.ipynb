```python
#Decorators 
# # A decorator is one of the functin which will provide additional functioality to the normal function
# # A decorator always takes normal function as parameter
# - syntax: 
#     def funcname(formalparam for normal function): #this is decorator definition/outer function definition
#             def function_name(): #inner function name
#                     block of statements
#                     provide addtional functionality to norma function
#                     .
#                     .
#                     return result of function_name
#             retun function_name

# - here function name is called decorator(outer function)
# - here function_name is called inner function            
```

```python
#exe1>
def square(mk): #decorator def
    def calculations():
        n=mk()
        res=n**2
        return n,res
    return calculations
def getval(): #this function def by mk
    return 5
#main program
calc=square(getval) #functin call takes normal function name as argument_decorator
n,res=calc()
print(f"suare of {n} is {res}")
```

```python
def square(gv):
    def calsqr():
        n=gv()
        res=n**2
        return n,res
    return calsqr
@square
def getval(): #this function def by mk 
    return float(input("Enter any number:"))

#main prog
num,result=getval() #normal functin call
print(f"square of {num} is = {result}")
```

```python
#ex3>
def squareroot(csqrt):
    def calsqrt():
        n,sqrv=csqrt()
        sqrtv=n**0.5
        return n,sqrv,sqrtv
    return calsqrt
def square(gv):
    def calsqr():
        n=gv()
        res=n**2
        return n,res
    return calsqr
@squareroot
@square
def getval():
    return float(input("Enter any number:"))
#main prog
num,sqrresult,sqrtresult=getval()
print(f"square of {num} is ={sqrresult}")
print(f"squareroot of {num} = {sqrtresult}")
    
```

```python
#ex4>
def cube(calsqrtval):
    def calcube():
        n,sqrval,sqrtval=calsqrtval()
        cbval=n**3
        return n,sqrval,sqrtval,cbval
    return calcube
def sqroot(csqrt):
    def calsqrtval():
        n,sqrt=csqrt()
        sqrtv=n**0.5
        return n,sqrt,sqrtv
    return calsqrtval
def square(gv):
    def calsqr():
        n=gv()
        res=n**2
        return n,res
    return calsqr
@cube
@sqroot
@square
def getval():
    return float(input("Enter any number:"))
num,sqrres,sqrtres,cbres=getval()
print(f"square of {num} is {sqrres}")
print(f"squaretoot of {num} is {sqrtres:.2f}")
print(f"cube of {num} is {cbres}")
```

```python
#variable scope
""" 
When a closure (inner function) is created i.e an inner function that captures variables from its enclosing scope(outer function) it retains access to those variables even after the outer function has finished execution this behaviour allows closure to remember and manipulate the values of variables from the eclosing scope (outer function)
"""
```

```python
#following is an example of closure with variables scope
def outer_function(x):
    y=10
    def inner_function(z):
        return x+y+z # x and y are captured from eclosing functin scope (outer func)
    return inner_function
#main prog
closure=outer_function(5)
res=closure(3)
print(res)
```

```python
#generators in python 
# -generator is one of the function
# - generator function always contains yield keyword 
# - if the function contains return statement then it called normal function here return statement of function can return more number of values if required
# - if the function contains yield keyword then it is called generator here yield statement returns the value only on demand and reduce memory spacce

# - syntax: 
#     def function_name(start,stop,step):
#             yield value

# - The yield keyword is used for giving the value back to function call from function call from function def and continue the function execution until condition becomes false
# - the advantage of generators over functions concept is that is save a lot of memory space in the case large sampling of data 
# - in otherwords function gives all the result at once and take more space where as generators gives one values at a time when programmer requested and takes minimized memory space
# - on the object of generator we can't perform indexing and slicing operators bcoz they supply the value only on demand.

# #within memory efficient programs using generators in python
# ---------------------------------------------------------------
# - when writting code in python when we are dealing with large amount of data it is mandatory to less memory space one way to do this is to use generators. generators are like special functions that help to save memory by processing data one at a time rather then all at once
# - the logic behind memory efficient function and python generators is to create functions that generate values on the fly. avoiding the need to store the entire data set in memory this is especially useful when dealing with large data sets.
```

```python
#ex1>
def kvrange(start,stop):
    while(start<=stop):
        yield start
        start=start+1
#main prog
r=kvrange(1,5) #function call return an obj of <class generator>
#to get the value from generator obj we use next()
for val in r:
    print(val)
    
```

```python
def kvrange(start,stop=1,step=1):
    if start>stop:
        stop=start
        start=1
    while(start<=stop):
        yield start
        start+=step

r1=kvrange(6)
print(next(r1))
print(next(r1))
print('-'*50)
for val in r1:
    print(val)
print('-'*50)

r2=kvrange(100,106)
print(next(r2))
print(next(r2))
print('-'*50)
for val in r2:
    print(val)
print('-'*50)
```

```python
#iterators in python
 
# why should we use:
# ------------------
# - in modern days we have a lot of data in our hands and handling this is huge amount of data creates problems for everyone who wants to do some sort of analysis with that data, so if we have ever struggled with handling huge amount of data and our machine running out of memory then we use concept of iterators in python
# - therefore rather than putting all the data in the memory in one step it would be better if we could work with it in bits or some small chunks dealing with only that data that is required at that moment as a result this would reduce the load on our computer memory tremendously and this is what exactly the iterators do
# - therefore you can use iterators to save a ton of memory as iterators dont compute this items when they are generated but only when they are called opon
# - iterator in python is an object that is used to iterate over iterable obj like lists,dicts,str, and sets but not applicable on non-iter obj like int,float,bool,complex,nontype
# - the iterator obj is initialized using the iter() method/function it uses the   next() for getting value from that iterator obj
# - here iter() is used for converting iterable obj into iterator obj
# - next() is used for obtaining next element of iterator object and if no next element found in iterator obj then we get an exception called stopiteration
# - on the obj of iterator we can't perform indexing & slicing operations coz they supply in value on demand

# example: 
# s="python"
# iterobj=iter(s)
# while(True):
#     try:
#        item=next(s)      #iterator by calling next()
#        print(item)
#     except StopIteration:     #exception will happen when iteration will over
#         break                      
```

```python
s="python"
iterobj=iter(s)
while(True):
    try:
       item=next(iterobj)      #iterator by calling next()
       print(item)
    except StopIteration:     #exception will happen when iteration will over
        break  
```
