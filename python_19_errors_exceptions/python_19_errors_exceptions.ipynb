```python
#           Types of Errors
# -------------------------------- 
# - The purpose of exception handling is that to develop or build robust(strong) applications
# - In real time, to develop any project, we develop, complie, execute variour programs During this process we get 3 types of errors they are
#             1>compile time errors
#             2>logical errors
#             3>runtime errors
#     1>compile time errors:
#     -------------------------
#     - These errors occurs during compilation time (.py---->pyc)
#     - These errors occur due to syntaxes are not followed during program development.
#     - These errors are solved by programmers during development time

#     example: SyntaxError, NameError, IndentationError

#     2>LogicalError:
#     ----------------------
#     - These errors are occurs during the runtime or execution time
#     - These erros occurs due to wrong or invalid reresentation of logic
#     - logicalErrors always gives wrong result
#     - These errors are solved by programmeres during development time

#     3> RuntimeError (implementatoinError):
#     -------------------------------------------
#     - These errors occurs during runtime or execution time
#     - These errors occurs due to wrong input/invalid input entered by end users or application users.
#     - when runtime errors occurs by default all the languages gives technical errors messages which are understandable by programmers but not to end-users this is not recommended process in real time
#     - According to industry standards it is recommended to display always user-friendly error messages for making the application robust by using exception handling.

#     **note: when the end-user enters valid input to the project gives successful result
#     - When the end user enters invalid input to the projec then the project must display user friendly error messages by using exceptionHandling. 

# #Building points:
# -----------------------
# - when the application user enters wrong or invalid input then we get runtime error (invalid input---->runtime error)
# - All runtime errors by default gives technical error messages (invalid input-->runtime error-->exception)

# #Definiton of Exception:
# --------------------------
# - every runtime error is called exception 
# - all invalid inputs are considered as exceptions
# - all exceptions by default gives technical error messages

# #Definition of exception handling:
# ---------------------------------------
# - The process of convertion technical error messages into user-friendly error messages is called exception handling
# - when the exception occurs in python programming , the following steps takes place internally
#     1>program execution abnormally terminated
#     2>pvm comes out of program flow
#     3>pvm by default generates technical error message

# - to do the above 1,2,3 pvm creates an object w.r.t an exception class
# - when exception occurs in python program then pvm creats an object of appropriate exception class
# - hence all exceptions are considered as objects of 'exceptionclass' (it is super class)
# ** note: all our errornames are called exception class name

# #types of exceptions in python:
# ----------------------------------
# - In python programming we have 2-types of exceptions they are
#     1>pre-defined or built-in exceptions
#     2>programmer or user or custom-defined exceptions

# 1>pre-defined or builtin exceptions:
# ----------------------------------------
# - predefined exceptions are those which are already developed by python lang developers and available inpython software and they are used by python lang programers for dealing with universal problams

# - some univeral problems they are:
#         1>invalid index (IndexError)
#         2>invalid conversion (ValueError)
#         3>invalid no.of args/operations (TypeError)
#         4>Division by Zero problems (ZeroDivisionError)

# - all the provide errors are called runtime errors and they called predefine exception class name

# 2> Programmer or userdefined or custom defined exceptions:
# ---------------------------------------------------------------
# - programmer or user custom defined exceptions are those which are developed by python programmers and they are available in python project and they are used by other theam members of same project and they are always deal with project and they are always deal with project common problems occuring itn the project.

# - some common problems are:
#         1>attempting to enter invalid username/password
#         2>attemptig to enter invalid PIN in atm applications 
#         3>attempting to withdraw more amount then existing balance of account
#         4>attempting to enter invalid name for people,places,products . ect....
```

```python
#example-1> Division

try:
    print("Program Execution started!")
    a=input("Enter first value:")
    b=input("Enter second value:")
    # convert a,b into floattype
    x=float(a)      #exception generated statement -- value error
    y=float(b)

    z=x/y   # exception generated statement -- zerodivision error

except ZeroDivisionError:
    print("Dont Enter zero for denominator")
except ValueError:
    print("Dont enter alnum,strs and symbols")
else:
    print("____ElseBlock____")
    print("first value=",x)
    print("second value=",y)
    print("Division=",z)
finally:
    print("____finally block_____")
    print("Program execution ended")
```

```python
#example-3>
a=input("Enter first value:")
b=input("Enter second value:")

try:
    x=float(a)
    y=float(b)
    z=x/y

except ZeroDivisionError as z:
    print("exception message:",z)
except ValueError as v:
    print("Exception message",v)
else:
    print(f"division of {x} and {y}={z}")
    
finally:
    print("Program executed completed")
```
