```python
# import sys
while True:
    print("Arithmetic operations")
    print('-'*50)
    print("1.Addition\n2.Substraction\n3.Multiplication\n4.Division\n5.Floor Division\n6.Modulo Division\n7.Exponential\n8.exit")
    print("-"*50)
    ch=int(input("Enter a number"))
    match(ch):
        case 1:
            print("Enter two values for addition")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a+b}")
        case 2:
            print("Enter two values for substraction")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a-b}")
        case 3:
            print("Enter two values for Multiplication")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a*b}")
        case 4:
            print("Enter two values for Division")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a/b}")
        case 5:
            print("Enter two values for Floor Division")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a//b}")
        case 6:
            print("Enter two values for Modulo Division")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a%b}")
        case 7:
            print("Enter two values for Exponential")
            a,b=float(input("Enter value for a:")),float(input("Enter value of b:"))
            print(f"the sum of {a},{b} is {a**b}")
        case 8:
            print("thanx for using the program")
            break
            # sys.exit()
        case _:
            print(f"you have entered {ch} is invalid input")
print("Execution completed")
    
```

```python
#area of different figures
con=1
while con:
    print("\n")
    print("choose any geometric figure to find its area")
    print("c.circle\nr.rectangle\ns.square\nt.triangle\ne.exit")
    print("\n")
    ch=input("choose one figure:")
    match (ch):
        case 'c'|'C':
            rad=float(input("Enter radius of circle:"))
            print(f"area of circle for radius {rad} is {(3.141*(rad*rad))}")
        case 'r'|'R':
            leng=float(input("Enter length of rectangle:"))
            wid=float(input("Enter width of rectangle:"))
            print(f"area of rectangle for  {leng,wid} is {(leng*wid)}")
        case 's'|'S':
            sid=float(input("Enter side of square:"))
            print(f"area of square for a side {sid} is {sid*sid}")
        case 't'|'T':
            bas=float(input("Enter base of triangle:"))
            hei=float(input("Enter height of triangle:"))
            print(f"area of triangle with base and height {bas,hei} is {0.5*bas*hei}")
        case 'e'|'E':
            print("you are terminated this program")
            # break
            con=0
        case _:
            print("you pic wrong option")
```

```python
#lcm and hcf/GCD of two numbers
a=int(input("enter value of a:"))
b=int(input('enter value of b:'))

if a<0 or b<0:
    print('invalid values')
else:
    oa=a
    ob=b
    while True:
        r=a%b
        if r==0:
            gcd=b
            print(f'GCD is {b}')
            break
        a=b
        b=r
    print(f'gcd of {oa},{ob}={gcd}')
    print(f"lcm of {oa},{ob}={(oa*ob)/gcd}")
```
